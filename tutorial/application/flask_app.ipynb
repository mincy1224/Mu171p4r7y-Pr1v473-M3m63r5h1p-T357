# Flask Deployment

The ``application/`` directory contains a reference Flask implementation
that wraps the mpmt package behind HTTP endpoints.

**Three servers** (Leader P0, Helper A P1, Helper B P2) each run a
Flask process.  A SetHolder CLI client communicates with them via HTTP.


## Server Startup

Start each server in its own terminal.  ``--prev-port`` and
``--next-port`` define the persistent Rep3 ring.

```bash
python application/server_app.py --party 0 --http-port 5000 --prev-port 14000 --next-host 127.0.0.1 --next-port 14001
python application/server_app.py --party 1 --http-port 5001 --prev-port 14001 --next-host 127.0.0.1 --next-port 14002
python application/server_app.py --party 2 --http-port 5002 --prev-port 14002 --next-host 127.0.0.1 --next-port 14000
```


## SetHolder CLI

```bash
python application/set_holder_client.py join --elements alice,bob,charlie
python application/set_holder_client.py update --elements alice,dave
python application/set_holder_client.py quit
```


## HTTP API

| Method | Endpoint | Request | Response |
|--------|----------|---------|----------|
| POST | ``/api/v1/reserve`` | token, action | ``{"status":"ok","port":N}`` |
| POST | ``/api/v1/connect`` | token, action, port | ``{"status":"ok"}`` |
| POST | ``/api/v1/aggregate`` | (none) | ``{"status":"ok"}`` |
| POST | ``/api/v1/query`` | element, port | ``{"status":"ok"}`` |

The application layer owns port allocation and token management.  The
mpmt package itself has no HTTP dependency — it works with any transport
layer (Flask, gRPC, direct TCP).
