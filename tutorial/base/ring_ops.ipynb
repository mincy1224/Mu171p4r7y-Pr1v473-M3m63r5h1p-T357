# Ring Operations

Scalar arithmetic in Z_{2^{ELL}}.  All operations are local (no network),
provided by the ``mpmt`` C++ extension.

- ``ring_add(ell, a, b)`` — (a + b) mod 2^ell
- ``ring_sub(ell, a, b)`` — (a - b) mod 2^ell
- ``ring_mul(ell, a, b)`` — (a * b) mod 2^ell
- ``ring_mod(ell, val, mod)`` — val % mod
- ``ring_mask(ell)`` — (1 << ell) - 1
- ``ring_rand(ell)`` — secure random element
- ``hash_aes_dm(preimage, key, ell)`` — AES-DM hash
- ``get_key_128bits()`` — random 16-byte key


```python
import mpmt

x = mpmt.ring_add(ell=4, a=3, b=5)
assert x == 8

y = mpmt.ring_sub(ell=8, a=10, b=20)
assert y == 246

z = mpmt.ring_mul(ell=4, a=3, b=5)
assert z == 15

r = mpmt.ring_mod(ell=14, val=42, mod=7)
assert r == 0

mask = mpmt.ring_mask(ell=4)
assert mask == 15

print("ring ops: OK")
```

## AES-DM Hashing

Used for Bloom filter construction.  Each hash function is parameterized
by a 128-bit seed; the output is an index in Z_{2^{ell}}.


```python
import mpmt

seed = mpmt.get_key_128bits()
assert len(seed) == 16

h = mpmt.hash_aes_dm(preimage=b"alice", key=seed, ell=14)
assert 0 <= h < (1 << 14)

k1 = mpmt.get_key_128bits()
k2 = mpmt.get_key_128bits()
assert k1 != k2

print("hash: OK")
```
