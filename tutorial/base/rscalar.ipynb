# $\mathbb{Z}_{2^{\ell}}$ Scalar Arithmetic

For consistency and controllability of Python-side semantics, `mpmt` wraps ring operations in C++. API ergonomics (e.g., operator overloading) are not yet considered; the API style mirrors that of vector operations.

## Basic Arithmetic

| Function | Signature | Semantics |
|------|------|------|
| `ring_add` | `(ell, a, b)` | `(a + b) mod 2^ell` |
| `ring_sub` | `(ell, a, b)` | `(a - b) mod 2^ell` |
| `ring_mul` | `(ell, a, b)` | `(a * b) mod 2^ell` |
| `ring_mod` | `(ell, val, mod)` | `val % mod`; `mod` must not be 0 |
| `ring_mask` | `(ell)` | Returns `(1 << ell) - 1`, i.e., the low `ell`-bit all-ones mask |

### Parameter Validation

All ring operations validate their inputs:
- `ell`: must be in `[1, 31]`; out-of-range raises `ValueError`
- `a`, `b`, `val`, `mod`: must be within `Z_{2^ell}` (`≤ mask`); out-of-range raises `ValueError`.

```python
import mpmt

# Addition
n1 = 7
n2 = 5
ell = 3
res = mpmt.ring_add(ell=ell, a=n1, b=n2)
print(f"{n1} + {n2} = {res} mod {2}^{ell}")

# Subtraction
res = mpmt.ring_sub(ell=ell, a=n1, b=n2)
print(f"{n1} - {n2} = {res} mod {2}^{ell}")

# Multiplication
res = mpmt.ring_mul(ell=ell, a=n1, b=n2)
print(f"{n1} * {n2} = {res} mod {2}^{ell}")

# Modulo
res = mpmt.ring_mod(ell=ell, val=n1, mod=n2)
print(f"{n1} % {n2} = {res} on Z_[2^{ell}]")
```

Inputs exceeding the modulus range raise ValueError

```python
n_big = 8
try:
    mpmt.ring_add(ell=ell, a=n_big, b=n1) 
except ValueError as e:
    print(f"ValueError: {e}")
```

$\mathbb{Z}_\ell$ Mask

```python
mask = mpmt.ring_mask(ell=ell)
print(f"mask({ell}) = 0b{mask:b}")
```

## Random Numbers

| Function | Signature | Semantics |
|------|------|------|
| `ring_rand` | `(ell)` | Cryptographically secure random element, `1 ≤ ell ≤ 31`; random tape sourced from libsodium |

Returns a random integer in `[0, 2^ell)`.

```python
r = mpmt.ring_rand(ell=ell)
print(f"random element on Z_[2^{ell}] = {r}")
```
