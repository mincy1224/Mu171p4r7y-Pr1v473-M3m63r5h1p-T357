# ShrRep3ShareVec / ShrRep3ShareScalar — RSS3 Share Containers

In RSS3 (3-party replicated secret sharing), each party holds **two** shares.
These are purely local data structures — no network communication. Actual shared values are produced by protocol methods such as `ShrRep3.share_vector` / `recv_vector_share`.

While Python tuples or pairs would suffice, these are wrapped in C++ for strict consistency across project layers.

## ShrRep3ShareScalar — Scalar Share

Fields (read-write):

| Field | Description |
|------|------|
| `this_share` | First RSS component |
| `nxt_share` | Second RSS component |

```python
import mpmt

ss = mpmt.ShrRep3ShareScalar()
ss.this_share = 7
ss.nxt_share  = 2
print(f"this_share = {ss.this_share}, nxt_share = {ss.nxt_share}")
```

## ShrRep3ShareVec — Vector Share

`ShrRep3ShareVec(ell)(size)` — Two-stage construction, `ell ∈ [1, 6]`.

Fields:

| Field | Type | Description |
|------|------|------|
| `this_share` | `Rvector(ell)` | First RSS component |
| `nxt_share` | `Rvector(ell)` | Second RSS component |
| `size` | `int` | Vector length (read-only) |

Setting `this_share` / `nxt_share` validates size equality — the two components must have the same length.

```python
import mpmt

SV = mpmt.ShrRep3ShareVec(ell=4)
sv = SV(size=50)
print(f"size = {sv.size}")

# Read/write components
sv.this_share[0] = 7
sv.nxt_share[0]  = 2
print(f"this_share[0] = {sv.this_share[0]}, nxt_share[0] = {sv.nxt_share[0]}")

# Setting unequal-length components raises ValueError
Rv = mpmt.Rvector(ell=4)
bad = Rv(30)
try:
    sv.this_share = bad
except ValueError as e:
    print(f"ValueError: {e}")
```
