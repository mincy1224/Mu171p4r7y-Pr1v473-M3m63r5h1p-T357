# ShrRep3ShareVec — RSS3 Share Vector

Each RSS3 party holds **two** components: ``this_share`` and ``nxt_share``.
Creating a ShareVec is local — no network communication.  Actual shares
are produced by ``ShrRep3.share_vector / recv_vector_share``.


```python
import mpmt

SV = mpmt.ShrRep3ShareVec(ell=4)
sv = SV(size=50)
assert sv.size == 50

sv.this_share[0] = 7
sv.nxt_share[0]  = 2
assert sv.this_share[0] == 7
assert sv.nxt_share[0]  == 2

print("sharevec: OK")
```
