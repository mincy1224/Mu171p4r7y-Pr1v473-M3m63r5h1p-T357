# Channel — Persistent TCP Channel

Wraps ``emp::NetIO``.  Two forms:
- ``Channel(port)`` — server: listen, block until a client connects
- ``Channel(host, port)`` — client: connect, retries until ready

One side listens; the other connects.  The code below requires
multi-party coordination.


```python
# This cell requires multi-party setup and will not run in a notebook

from mpmt.channels import Channel, _build_rep3_channels

# Point-to-point
ch_srv = Channel(port=14000)
ch_cli = Channel(host="127.0.0.1", port=14000)

# Rep3 ring (3-party, thread-free)
ch = _build_rep3_channels(
    prev_port=14000, next_host="127.0.0.1", next_port=14001,
    party_id=0,
)

# DPF star
from mpmt.channels import _build_dpf_channels_dealer, _build_dpf_channels_evaluator
ch_d = _build_dpf_channels_dealer(eval0_port=18000, eval1_port=18001)
ch_e = _build_dpf_channels_evaluator(dealer_host="127.0.0.1", dealer_port=18000)
```

## Channel Methods

- ``ch.acquire()`` — flush, reset counters, return ``IOChannel*``
- ``ch.flush()`` — flush send buffer
- ``ch.send(data)`` — send ``bytes`` or ``bytearray``
- ``ch.recv(buf)`` — receive exactly ``len(buf)`` bytes
