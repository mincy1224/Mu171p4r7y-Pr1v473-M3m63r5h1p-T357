## Protocol Usage

- **SetHolder → Leader**: ELL=1 vector mode (BF share components)
- **Querier → Helpers**: ELL=ell_add2 scalar mode (hash indices)
- **Helpers → Querier**: ELL=ell_query scalar mode (dot shares)
- **Leader → Querier**: ELL=ell_query scalar mode (ET share)
